## Loss Functions(MSE,MAE, HUBER)

```python
import matplotlib.pyplot as plt
import numpy as np
```

### Mean-Squared Error(MSE)
- Used in Regression Networks, AutoEncoders, Reconstruction Losses
- Smooth and Easy Optimization(**PROS**)
- Sensitive to Outliers(**CONS**)
- Doesn't work good for Image Reconstruction(**CONS**)

![mse.png](./imgs/mse.png)

```python
def mse_loss(y_true, y_pred):
    r = y_true - y_pred
    loss = 0.5 * np.mean(r ** 2)
    grad = -r / y_true.shape[0]
    return loss, grad
```

```python
r = np.linspace(-3,3,400)
plt.plot(r, 0.5*r**2)
plt.title("MSE Loss")
plt.xlabel("Error(r)")
plt.ylabel("Loss")
plt.show()
```
